# Peak fitting example

Demonstrates the `PeakFit` class and area calculation.

```python
%matplotlib inline
import numpy as np
import matplotlib.pyplot as plt
from wara import peaksearch as ps
from wara import peakfit as pf
from wara import file_reader
```

Load the spectrum and set resolution parameters.

```python
file = "data/test_data_cebr.csv"
spect = file_reader.read_csv(file)

fwhm_at_0 = 1.0
ref_fwhm = 31
ref_x = 1220
min_snr = 5
```

Run the peak search.

```python
search = ps.PeakSearch(spect, ref_x, ref_fwhm, fwhm_at_0, min_snr=min_snr)
search.plot()
plt.show()
```

Fit peaks in the range [1250, 1600] channels using a linear background.

```python
bkg = "poly1"
xrange = [1250, 1600]
fit = pf.PeakFit(search, xrange, bkg=bkg)
fit.plot()
plt.show()
print(fit.peak_info)
```

Compare the peak area from `PeakFit` against an independent Gaussian integration.

```python
def gaussian(x, A, mu, sig):
    return A / np.sqrt(2 * np.pi * sig**2) * np.exp(-((x - mu)**2) / (2 * sig**2))

res = fit.fit_result
xg = fit.x_data
Ag  = res.best_values["g1_amplitude"]
mug = res.best_values["g1_center"]
sig = res.best_values["g1_sigma"]

print(f"Area from PeakFit:          {Ag:.1f}")
print(f"Area from Gaussian integral: {gaussian(xg, Ag, mug, sig).sum():.1f}")
```
