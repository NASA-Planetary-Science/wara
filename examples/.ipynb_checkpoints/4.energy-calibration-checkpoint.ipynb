# Energy calibration example

Calibrates a CeBr3 spectrum using Co-60 and Cs-137 gamma lines.

```python
%matplotlib inline
import matplotlib.pyplot as plt
from wara import peaksearch as ps
from wara import peakfit as pf
from wara import energy_calibration as ecal
from wara import file_reader
```

Load the spectrum and run a peak search.

```python
file = "data/test_data_cebr.csv"
spect = file_reader.read_csv(file)

fwhm_at_0 = 1.0
ref_x = 1315
ref_fwhm = 42
search = ps.PeakSearch(spect, ref_x, ref_fwhm, fwhm_at_0, min_snr=4)
```

Fit peaks in the range [680, 1600] channels.

```python
xrange = [680, 1600]
bkg = "poly1"
fit = pf.PeakFit(search, xrange=xrange, bkg=bkg)
fit.plot()
plt.show()
print(fit.peak_info)
```

Map fitted peak centroids to known gamma energies (Cs-137: 662 keV, Co-60: 1173.2 and 1332.5 keV) and build the calibration.

```python
peak_info = fit.peak_info
centroids = [peak_info[0]['mean'], peak_info[2]['mean'], peak_info[3]['mean']]
energies_kev = [662, 1173.2, 1332.5]

cal = ecal.EnergyCalibration(
    mean_vals=centroids, erg=energies_kev,
    channels=spect.channels, n=1, e_units='keV'
)
cal.plot()
plt.show()
```

Apply the calibration to the spectrum and plot.

```python
spect.apply_calibration(cal)
spect.plot()
plt.title("Calibrated spectrum")
plt.show()
```
