# Peak search example — CeBr3 detector

Demonstrates the `PeakSearch` class on a CeBr3 spectrum.

```python
%matplotlib inline
import matplotlib.pyplot as plt
from wara import peaksearch as ps
from wara import file_reader
```

Load the spectrum.

```python
file = "data/test_data_cebr.csv"
spect = file_reader.read_csv(file)
```

Set the resolution model parameters (all in channels).

- `fwhm_at_0`: FWHM extrapolated to channel 0
- `ref_x`: reference channel for the FWHM measurement
- `ref_fwhm`: measured FWHM at `ref_x`

```python
fwhm_at_0 = 1
ref_fwhm = 20
ref_x = 420
```

Run the peak search and plot the kernel, the detected peaks, and the spectral components.

```python
search = ps.PeakSearch(spect, ref_x, ref_fwhm, fwhm_at_0, min_snr=5, method="km")
search.plot_kernel()
plt.show()
search.plot()
plt.show()
search.plot_components()
plt.show()
```

Search within a restricted channel range and inspect metadata.

```python
search1 = ps.PeakSearch(
    spect, ref_x, ref_fwhm, fwhm_at_0, min_snr=5, xrange=[1200, 1600], method="km"
)
search1.plot(snrs="off")
plt.show()
print(search1.metadata())
```

List peaks found within a specific channel range.

```python
search.peaks_in_range(x_min=0, x_max=300)
```
